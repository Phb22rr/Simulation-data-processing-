```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import csv
import random
#import g4ppyy as g4
#from g4ppyy.SI import mm, cm, m, eV, MeV, GeV, deg, twopi
```

```python
fieldnames = ['radiation_number', 'incident_radiation_energy', 'energy_deposited', 'max_pen']

Alpha_dict = {key: [] for key in fieldnames}
Electron_dict = {key: [] for key in fieldnames}
Gamma_dict = {key: [] for key in fieldnames}
Neutron_dict = {key: [] for key in fieldnames}
Proton_dict = {key: [] for key in fieldnames}

with open('Data --- simulations/10keV/Alpha10keV.csv', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        Alpha_dict['radiation_number'].append(row['radiation_number'])
        Alpha_dict['incident_radiation_energy'].append(float(row['incident_radiation_energy']))
        Alpha_dict['energy_deposited'].append(float(row['energy_deposited']))
        Alpha_dict['max_pen'].append(float(row['max_pen']))

with open('Data --- simulations/10keV/Electron10keV.csv', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        Electron_dict['radiation_number'].append(row['radiation_number'])
        Electron_dict['incident_radiation_energy'].append(float(row['incident_radiation_energy']))
        Electron_dict['energy_deposited'].append(float(row['energy_deposited']))
        Electron_dict['max_pen'].append(float(row['max_pen']))

with open('Data --- simulations/10keV/Gamma10keV.csv', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        Gamma_dict['radiation_number'].append(row['radiation_number'])
        Gamma_dict['incident_radiation_energy'].append(float(row['incident_radiation_energy']))
        Gamma_dict['energy_deposited'].append(float(row['energy_deposited']))
        Gamma_dict['max_pen'].append(float(row['max_pen']))

with open('Data --- simulations/10keV/Neutron10keV.csv', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        Neutron_dict['radiation_number'].append(row['radiation_number'])
        Neutron_dict['incident_radiation_energy'].append(float(row['incident_radiation_energy']))
        Neutron_dict['energy_deposited'].append(float(row['energy_deposited']))
        Neutron_dict['max_pen'].append(float(row['max_pen']))

with open('Data --- simulations/10keV/Proton10keV.csv', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        Proton_dict['radiation_number'].append(row['radiation_number'])
        Proton_dict['incident_radiation_energy'].append(float(row['incident_radiation_energy']))
        Proton_dict['energy_deposited'].append(float(row['energy_deposited']))
        Proton_dict['max_pen'].append(float(row['max_pen']))
```

```python
for df in [Alpha_df, Electron_df, Gamma_df, Neutron_df, Proton_df]:
    z_list = df.z.values.tolist()
    #print(np.shape(z_list))
    #print(z_list)
    edep_list = df.edep.values.tolist()
    #print(np.shape(edep_list))
    #print(edep_list)

    total_incident_energy = np.sum(df.groupby("eid")["gen_gamma_energy"].first())#np.sum(df.true_gamma_energy)

    sorted_indices = np.argsort(z_list)
    z_list = np.array(z_list)[sorted_indices]
    z_list = z_list - 5
    edep_list = np.array(edep_list)[sorted_indices]
    edep_list = (edep_list / total_incident_energy)*100

    cum_sum_edep = np.cumsum(edep_list)
    #print(cum_sum_edep[-1])

    if df is Alpha_df:
        plt.plot(z_list, cum_sum_edep, label="Alpha 10 keV")
    elif df is Electron_df:
        plt.plot(z_list, cum_sum_edep, label="Electron 10 keV")
    elif df is Gamma_df:
        plt.plot(z_list, cum_sum_edep, label="Gamma 10 keV")
    elif df is Neutron_df:
        plt.plot(z_list, cum_sum_edep, label="Neutron 10 keV")
    elif df is Proton_df:
        plt.plot(z_list, cum_sum_edep, label="Proton 10 keV")

    plt.xlabel("Penetration depth (mm)")
    plt.ylabel("Percentage of total cumulitive energy deposited (%)")
    plt.legend(loc="lower left")
    plt.xscale('log')
    plt.yscale('log')
    plt.xlim([1*10**(-6), 1])
    plt.ylim([0.01, 100])
    #plt.ylim(72.03, 72.0375)
    #plt.xlim(7.9*10**-4 ,9 * 10**-4)
    plt.title("10 keV radiation cumulative energy deposition")
    plt.tick_params(axis='both' , direction='in', which='both')
    #plt.show()
plt.show()
```

```python
#penetration depth as defined in that one paper as 99% of energy deposited
taper_check_number = - 200
taper_check_percentage = 0.005

for df in [Alpha_df, Electron_df, Gamma_df, Neutron_df, Proton_df]:
    if df is Alpha_df:
        name = "Alpha"
    elif df is Electron_df:
        name = "Electron"
    elif df is Gamma_df:
        name = "Gamma"
    elif df is Neutron_df:
        name = "Neutron"
    elif df is Proton_df:
        name = "Proton"

    z_list = df.z.values.tolist()
    #print(np.shape(z_list))
    #print(z_list)
    edep_list = df.edep.values.tolist()
    #print(np.shape(edep_list))
    #print(edep_list)

    total_incident_energy = np.sum(df.groupby("eid")["gen_gamma_energy"].first())#np.sum(df.true_gamma_energy)

    sorted_indices = np.argsort(z_list)
    z_list = np.array(z_list)[sorted_indices]
    z_list = z_list - 5
    z_list = z_list.tolist()
    edep_list = np.array(edep_list)[sorted_indices]
    edep_list = (edep_list / total_incident_energy)*100

    cum_sum_edep = np.cumsum(edep_list)

    pen_depth_list = []

    if cum_sum_edep[-1] < 99:
        print(f"Warning: Cumulative energy deposition for {name} does not reach 99%. Maximum is {cum_sum_edep[-1]}%.")

        if cum_sum_edep[-1] <= taper_check_percentage * cum_sum_edep[-1] + cum_sum_edep[taper_check_number]:
            print(f"Radiaiton tapers off.")

            assume_100_percent = 100/cum_sum_edep[-1]
            scaled_cum_sum_edep = assume_100_percent * cum_sum_edep
            for i, val in enumerate(scaled_cum_sum_edep):
                if val>99:
                    hello = z_list[i]
                    pen_depth_list.append(hello)

            pen_depth_list = np.array(pen_depth_list)
            pen_depth = np.min(pen_depth_list)

            print(f"Approximated penetration depth for 10keV {name} = {pen_depth}mm")

        elif cum_sum_edep[-1] > taper_check_percentage * cum_sum_edep[-1] + cum_sum_edep[taper_check_number]:
            print("Radiation does not taper off. Penetration depth >1mm")
        else:
            print("Error in tapering")

    elif cum_sum_edep[-1] > 99:
        for i, val in enumerate(cum_sum_edep):
            if val>99:
                pen_depth_list.append(z_list[i])

        pen_depth_list = np.array(pen_depth_list)
        #print(pen_depth_list)
        pen_depth = np.min(pen_depth_list)
        print(f"True penetration depth for 10keV {name} = {pen_depth}mm")

    else:
        print("Error. Somehow the final item of cum_sum_edep is not > or < to 99")
    print("~")
```
